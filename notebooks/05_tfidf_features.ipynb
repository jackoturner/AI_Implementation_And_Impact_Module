# TF-IDF Text Features

Machine learning models require numerical features. This notebook converts review text into TF-IDF features using only the training data, then applies the fitted transformation to the unseen test data.

```python
from pathlib import Path

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split

MODEL_SAMPLE_PATH = Path(
    "../data/processed/video_games_model_sample.csv"
)

RANDOM_STATE = 42

reviews = pd.read_csv(
    MODEL_SAMPLE_PATH,
    keep_default_na=False,
)

X = reviews["reviewText"]
y = reviews["sentiment"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y,
)
```

```python
tfidf_vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=3,
    max_df=0.95,
    max_features=15_000,
    sublinear_tf=True,
)

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

print(f"Training feature matrix: {X_train_tfidf.shape}")
print(f"Testing feature matrix: {X_test_tfidf.shape}")
print(f"Vocabulary size: {len(tfidf_vectorizer.vocabulary_):,}")
```

```python
feature_names = tfidf_vectorizer.get_feature_names_out()

print("First 30 TF-IDF features:")
print(feature_names[:30])
```

## TF-IDF Feature Conclusion

The TF-IDF transformation created 15,000 numerical text features for each review. The vectoriser was fitted only on the 80,000 training reviews, then applied unchanged to the 20,000 test reviews.

Both individual words and two word phrases were included. Numeric terms were retained because numbers can be meaningful in video game reviews, for example when referring to prices, playtime, ratings or game versions.
