# Initial Video Games Review Data Check

Notebook loads a small preveiw of the Video Games 5-core dataset to try to understand its structure before any cleaning or modelling.

```python
from pathlib import Path
import gzip
import json

import pandas as pd

RAW_DATA_PATH = Path("../data/raw/Video_Games_5.json.gz")
PREVIEW_ROWS = 10_000

records = []

with gzip.open(RAW_DATA_PATH, "rt", encoding="utf-8") as file:
    for index, line in enumerate(file):
        if index >= PREVIEW_ROWS:
            break
        records.append(json.loads(line))

reviews_preview = pd.DataFrame(records)

print(f"Preview rows: {reviews_preview.shape[0]:,}")
print(f"Columns: {reviews_preview.shape[1]}")
reviews_preview.head()
```
