# Train-Test Split

This notebook creates stratified training and testing sets from the 100,000-row modelling sample.

```python
from pathlib import Path

import pandas as pd
from sklearn.model_selection import train_test_split

MODEL_SAMPLE_PATH = Path(
    "../data/processed/video_games_model_sample.csv"
)

RANDOM_STATE = 42

reviews = pd.read_csv(MODEL_SAMPLE_PATH)

X = reviews["reviewText"]
y = reviews["sentiment"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y,
)

print(f"Training reviews: {len(X_train):,}")
print(f"Testing reviews: {len(X_test):,}")
```

```python
sentiment_order = ["negative", "neutral", "positive"]

split_distribution = pd.DataFrame({
    "training_percentage": (
        y_train.value_counts(normalize=True)
        .reindex(sentiment_order)
        .mul(100)
    ),
    "testing_percentage": (
        y_test.value_counts(normalize=True)
        .reindex(sentiment_order)
        .mul(100)
    ),
}).round(2)

split_distribution
```

## Split Conclusion

The 80/20 stratified split preserved the sentiment label proportions in both datasets. A fixed random state of 42 was used so that the split can be reproduced.

The positive class remains dominant, so accuracy alone will not be sufficient for comparing models. Macro F1-score, per-class precision and recall, and confusion matrices will be used during evaluation.
